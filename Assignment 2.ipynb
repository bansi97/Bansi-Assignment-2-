```python
#Q2. You are given a list that contains some duplicate items. Remove the duplicates by converting the list to a set. 
#Final output should be in list
shopping_list = ["apple", "banana", "apple", "orange", "banana", "grape"]

# Convert the list to a set to remove duplicates
unique_items = set(shopping_list)

# Convert the set back to a list
shopping_list = list(unique_items)

# Display the final list
print(shopping_list)
```

```python
#Q3. You are managing a list of students enrolled in two different courses. 
#You need to perform various set operations to understand the student enrollment. 
#Task: Create two sets: one for students in "Course A" and one for students in "Course B". 
#Find the students who are in "Course A" but not in "Course B" (difference).
#Find students who are only in one of the two courses (symmetric difference).

course_a = {"John", "Alice", "Bob", "David"} 
course_b = {"Alice", "Eve", "Charlie", "David"}

# Find students in Course A but not in Course B
difference = course_a.difference(course_b)

# Find students who are in only one of the two courses
symmetric_difference = course_a.symmetric_difference(course_b)

# Display the results
print("Students in Course A but not in Course B:", difference)
print("Students only in one of the two courses:", symmetric_difference)
```

```python
#Q6. Set Comparison
A = {1, 2, 3, 4, 5}
B = {4, 5, 6, 7}

# Find common elements
common = A.intersection(B)

# Find elements only in Set A
only_a = A.difference(B)

# Find elements only in Set B
only_b = B.difference(A)

# Display the results
print("Common elements:", common)
print("Elements only in Set A:", only_a)
print("Elements only in Set B:", only_b)
```

```python
#Q7. Given the tuple my_tuple = (1, 2, 3, 4, 5), remove the element 3 from the tuple by first converting it to a list,
# and then print the resulting tuple.
my_tuple = (1, 2, 3, 4, 5)

# Convert tuple to a list
my_list = list(my_tuple)

# Remove the element 3
my_list.remove(3)

# Convert the list back to a tuple
my_tuple = tuple(my_list)

# Print the resulting tuple
print(my_tuple)
```
